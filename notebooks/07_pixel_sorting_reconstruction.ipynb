# Phase 9: Exact Pixel Sorting (The Generalization Algorithm)
This notebook demonstrates the beautiful 1-to-1 pixel sorting algorithm.

It takes **EVERY SINGLE PIXEL** from the Source Image (e.g. Icarus), and perfectly rearranges them to form the Target Image (e.g. Conor McGregor). 
This mathematically guarantees that the final image has 100% of the exact color palette of Icarus, but perfectly reconstructed as Conor!

```python
import sys
import os
import matplotlib.pyplot as plt
from PIL import Image

# Add src to path
sys.path.append(os.path.abspath('..'))

from src.models.pixel_sorter import PixelSorterEngine
```

```python
# 1. Initialize the Engine
engine = PixelSorterEngine()

# 2. Define the Target (Shape) and Source (Colors)
# You can change these to ANY two images you want!
target_path = '../data/content_images/mcgregor.jpg'  # The Shape/Shadows
source_path = '../data/style_images/icarus.jpg'      # The Pixels/Colors

# 3. Run the Engine! 
# We will use 400x400 (160,000 pixels). This runs instantly on CPU.
final_image = engine.run(target_path, source_path, output_size=(400, 400))
```

```python
# 4. Display the Magic
fig, axs = plt.subplots(1, 3, figsize=(20, 8))

axs[0].imshow(Image.open(target_path).resize((400, 400)))
axs[0].set_title('Target Shape')
axs[0].axis('off')

axs[1].imshow(Image.open(source_path).resize((400, 400)))
axs[1].set_title('Source Pixels (Colors)')
axs[1].axis('off')

axs[2].imshow(final_image)
axs[2].set_title('Exact Pixel Sorting Output')
axs[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Optional: Save to disk
os.makedirs('../outputs', exist_ok=True)
final_image.save('../outputs/mcgregor_pixelsorted.jpg')
print("Masterpiece saved to outputs folder!")
```
