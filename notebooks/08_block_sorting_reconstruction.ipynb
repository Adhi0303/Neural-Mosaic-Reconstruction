# Phase 9.1: 1-to-1 Block Sorting (The Ultimate Solution)
This is the perfect blend of the Patch Mosaic and the Exact Pixel Sorter.

It chops the Source image into large chunks (e.g., 20x20 pixels), so the red skirts and blue skies stay clustered together. 
Then, it perfectly maps the chunks 1-to-1 onto the Target image. Every single chunk is used exactly once! No more "muddy sand" effect!

```python
import sys
import os
import matplotlib.pyplot as plt
from PIL import Image

# Add src to path
sys.path.append(os.path.abspath('..'))

from src.models.block_sorter import BlockSorterEngine
```

```python
# 1. Initialize the Engine with a block size
# Set block_size=20 for a chunky "Rubik's Cube" look. 
engine = BlockSorterEngine(block_size=20)

# 2. Define the Target (Shape) and Source (Colors)
target_path = '../data/content_images/mcgregor.jpg'  # The Shape/Shadows
source_path = '../data/style_images/icarus.jpg'      # The Pixels/Colors

# 3. Run the Engine! 
final_image = engine.run(target_path, source_path, output_size=(400, 400))
```

```python
# 4. Display the Magic
fig, axs = plt.subplots(1, 3, figsize=(20, 8))

axs[0].imshow(Image.open(target_path).resize((400, 400)))
axs[0].set_title('Target Shape')
axs[0].axis('off')

axs[1].imshow(Image.open(source_path).resize((400, 400)))
axs[1].set_title('Source Blocks (Colors)')
axs[1].axis('off')

axs[2].imshow(final_image)
axs[2].set_title('Exact Block Sorting Output')
axs[2].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Optional: Save to disk
os.makedirs('../outputs', exist_ok=True)
final_image.save('../outputs/mcgregor_blocksorted.jpg')
print("Masterpiece saved to outputs folder!")
```
